```python
import json
import numpy as np

machine = "FLAMENIR"
fold_accs = []
fold_precisions = []
fold_recalls = []
fold_f1s = []

for idx in range(5):
    with open(f"history/category_classification/{machine}/smart_nir_classification_fold{idx+1}.json", "r") as file:
        history = json.load(file)
    max_idx = np.argmax(history['val_acc'])
    
    print(
        f"Fold {idx + 1} --- "
        f"Accuracy: {history['val_acc'][max_idx] * 100:.2f} | "
        f"Precision: {history['val_precision'][max_idx] * 100:.2f} | "
        f"Recall: {history['val_recall'][max_idx] * 100:.2f} | "
        f"F1: {history['val_f1'][max_idx] * 100:.2f}"
    )
    fold_accs.append(history['val_acc'][max_idx])
    fold_precisions.append(history['val_precision'][max_idx])
    fold_recalls.append(history['val_recall'][max_idx])
    fold_f1s.append(history['val_f1'][max_idx])

print(
    f"Average --- "
    f"Accuracy: {np.mean(fold_accs) * 100:.2f} | "
    f"Precision: {np.mean(fold_precisions) * 100:.2f} | "
    f"Recall: {np.mean(fold_recalls) * 100:.2f} | "
    f"F1: {np.mean(fold_f1s) * 100:.2f}"
)
```

```python
import json
import numpy as np
import os

machine = "FLAMENIR"

for substance in [
    'Thiamethoxam', 'Permethrin', 'Metalaxyl', 'Azoxystrobin',
    'Imidaclopird', 'Difenoconazole', 'Cypermethrin', 'Cyhalothrin',
    'Chlorantraniliprol', 'Chlopyrifos Methyl', 'Emamectin benzoate',
    'Chlorothalonil', 'Triadimefon', 'Cyantraniliprole', 'Flutolanil',
    'Indoxacarb', 'Abamectin', 'Propamocarb.HCL', 'Chlothianidin'
]:  
    print(f"===== {substance} =====")
    
    if len(os.listdir(f"history/substance_regression/stage1/{machine}/{substance}")) == 0:
        continue

    fold_accs = []
    fold_precisions = []
    fold_recalls = []
    fold_f1s = []

    for idx in range(5):
        with open(f"history/substance_regression/stage1/{machine}/{substance}/{substance}_fold_{idx+1}.json", "r") as file:
            history = json.load(file)
        max_idx = np.argmax(history['val_acc'])
        fold_accs.append(history['val_acc'][max_idx])
        fold_precisions.append(history['val_precision'][max_idx])
        fold_recalls.append(history['val_recall'][max_idx])
        fold_f1s.append(history['val_f1'][max_idx])

    print(
        f"Average --- "
        f"Accuracy: {np.mean(fold_accs) * 100:.2f} | "
        f"Precision: {np.mean(fold_precisions) * 100:.2f} | "
        f"Recall: {np.mean(fold_recalls) * 100:.2f} | "
        f"F1: {np.mean(fold_f1s) * 100:.2f}"
    )
```

```python
import json
import numpy as np
import os

machine = "OCEANFX"

for substance in [
    'Thiamethoxam', 'Permethrin', 'Metalaxyl', 'Azoxystrobin',
    'Imidaclopird', 'Difenoconazole', 'Cypermethrin', 'Cyhalothrin',
    'Chlorantraniliprol', 'Chlopyrifos Methyl', 'Emamectin benzoate',
    'Chlorothalonil', 'Triadimefon', 'Cyantraniliprole', 'Flutolanil',
    'Indoxacarb', 'Abamectin', 'Propamocarb.HCL', 'Chlothianidin'
]:  
    print(f"===== {substance} =====")
    try:
        if len(os.listdir(f"history/substance_regression/stage2/{machine}/{substance}")) == 0:
            print(
                f"Average --- "
                f"MSE: NaN | "
                f"MAE: NaN | "
                f"RMSE: NaN | "
                f"R2: NaN"
            )
            continue
        
        fold_mses = []
        fold_maes = []
        fold_rmses = []
        fold_r2 = []

        for idx in range(5):
            with open(f"history/substance_regression/stage2/{machine}/{substance}/smart_nir_regression_fold{idx+1}.json", "r") as file:
                history = json.load(file)
            min_idx = np.argmin(history['val_loss'])
            fold_mses.append(history['val_loss'][min_idx])
            fold_maes.append(history['val_mae'][min_idx])
            fold_rmses.append(history['val_rmse'][min_idx])
            fold_r2.append(history['val_r2'][min_idx])

        print(
            f"Average --- "
            f"MSE: {np.mean(fold_mses):.3f} | "
            f"MAE: {np.mean(fold_maes):.3f} | "
            f"RMSE: {np.mean(fold_rmses):.3f} | "
            f"R2: {np.mean(fold_r2) * 100:.2f}"
        )
    except:
        continue
```
